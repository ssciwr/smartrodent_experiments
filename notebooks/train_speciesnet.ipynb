# Generate dataset fit for Speciesnet

We need to make sure that train and validation datasets don't contain the same repetetive images and run the same backrounds repeatedly

```python
from smartrodent import SpeciesNet_Detector
from pathlib import Path
path = "/home/hmack/projects/smartrodent_experiments/datasets/full_dataset/filtered_kept/Apodemus agrarius/31091381.jpg"
path_to_labels = "./dummy"

detector_kwargs={"model_name": "kaggle:google/speciesnet/pyTorch/v4.0.3b/1",
                "crop": True,
                "country": None,
                "relpad": 0.0,
                "geofence": False,
                "components": "all"
                }

detector = SpeciesNet_Detector(
        name="boxed",
        batchsize=80,
        img_outname="boxed",
        conf=0.01,
        project=path_to_labels,
        **detector_kwargs,
    )

detector.detect([path, path, path], Path(path_to_labels) )
```
