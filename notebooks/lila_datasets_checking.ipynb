# Explore island conservation data

The LILA JSON is in COCO Camera Traps format: a dict with separate `images`, `annotations`, `categories` and `info` tables, so `pd.read_json` can't load it directly. Build one dataframe per table and merge them (one row per annotation; `bbox` is `[x, y, width, height]` in pixels).

```python
import json
import pandas as pd

path = "./datasets/LiLa_island_conservation/island_conservation_camera_traps_1.02.json"

with open(path) as f:
    data = json.load(f)

images = pd.DataFrame(data["images"]).rename(columns={"id": "image_id"})
annotations = pd.DataFrame(data["annotations"])
categories = pd.DataFrame(data["categories"]).rename(columns={"id": "category_id", "name": "category"})

df = (
    annotations
    .merge(categories, on="category_id", how="left")
    .merge(images, on="image_id", how="left")
)
df.head()

display(df["category"].value_counts())
```
